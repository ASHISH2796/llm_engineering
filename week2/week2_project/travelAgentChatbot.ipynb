```python
#import
import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
from fileReaderUtility import readSystemPrompt

#Load env - not usefull in our case as we use local model ollama
load_dotenv(override=True)
#openai_api_key = os.getenv('OPENAI_API_KEY')
openai_api_key = "ollama"

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")

#Ollama config setup
OLLAMA_MODEL="llama3.2"
OLLAMA_URL = "http://localhost:11434/v1"
ollama = OpenAI(api_key=openai_api_key,base_url=OLLAMA_URL)

#system prompt
system_prompt = readSystemPrompt()

#callback function/generator with streams
def chat(message, history):
    #payload
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]

    #stream call
    stream = ollama.chat.completions.create(model = OLLAMA_MODEL, messages = messages, stream = True)
    
    #response processing
    response =''
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response

    

#gradio UI config
gr.ChatInterface(fn=chat, type="messages").launch()
```
